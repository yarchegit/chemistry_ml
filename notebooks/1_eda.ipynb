```python
df_raw.head()
```

## Общая характеристика датасета

* Размер выборки и количество признаков позволяют строить достаточно сложные модели без явного риска переобучения, так как данных мало 
* Признаки представлены в основном числовыми дескрипторами и также тремя таргетами IC50, CC50 и SI.

```python
#Базовая очистка
na_counts = df_raw.isna().sum().sort_values(ascending=False)
na_counts.head(20)
```

```python
target_cols = ['IC50, mM', 'CC50, mM', 'SI']
df = df_raw.dropna(subset=target_cols).copy()
feature_na_ratio = (df.isna().sum() / len(df)).drop(target_cols)
high_na_features = feature_na_ratio[feature_na_ratio > 0.3].index
low_na_features = feature_na_ratio[feature_na_ratio > 0].index.difference(high_na_features)
df = df.drop(columns=high_na_features)


from sklearn.impute import SimpleImputer

feature_cols = df.columns.difference(target_cols)
imputer = SimpleImputer(strategy='median')
df[feature_cols] = imputer.fit_transform(df[feature_cols])
df.isna().sum().sum()
```

## Пропуски и удаление нерелевантных признаков

* Объекты с пропусками в таргетах удалены, так как они не пригодны для обучения моделей
* Признаки с долей пропусков выше порога -30 % - исключены как нерелевантные, оставшиеся значения заполнены медианой, что уменьшает смещение и сохраняет распределения

```python
df[target_cols].describe().T
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, col in zip(axes, target_cols):
    sns.histplot(df[col], kde=True, ax=ax)
    ax.set_title(f'Распределение {col}')
plt.tight_layout()
```

```python
plt.figure(figsize=(10, 4))
sns.boxplot(data=df[target_cols])
plt.title('Бокс для таргетов')
plt.show()
```

## Анализ таргетов и выбросов

* Распределения асимметричны, присутствуют значимые выбросы, что ок для био параметров
* Выбросы отражают реально существующие соединения с высокой или низкой активностью, поэтому они сохранены, а устойчивость моделей обеспечивается выбором алгоритмов и метрик

```python
corr_targets = df[target_cols].corr()
corr_targets
```

```python
corr_with_targets = df.corr(numeric_only=True)[target_cols].sort_values('IC50, mM', ascending=False)
corr_with_targets.head(15)
```

## Корреляционный анализ

* Между IC50, CC50 и SI наблюдается заметная корреляция, что отражает взаимосвязь эффективности и цитотоксичности соединений 
* Часть дескрипторов сильнее всего коррелируют с таргетами, их можно рассматривать как основное ядро признакового пространства.
